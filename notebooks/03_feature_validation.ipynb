```python
import pandas as pd
import numpy as np
import sys
import os

sys.path.insert(0, os.path.abspath('..'))

from src.features import extract_features, get_feature_names

print("Importações OK!")
print(f"Features esperadas: {len(get_feature_names())}")
```

```python
# Carregar CSV e pegar uma música de teste
df = pd.read_csv('../data/features_30_sec.csv')

# Pegar a primeira música do dataset
sample_row = df.iloc[0]
filename = sample_row['filename']
genre = sample_row['label']

print(f"Música de teste: {filename}")
print(f"Gênero: {genre}")

# Valores do CSV para comparação
csv_features = sample_row.drop(['filename', 'length', 'label']).values
print(f"\nPrimeiros 5 valores do CSV:")
print(csv_features[:5])
```

```python
# Caminho do arquivo de áudio
audio_path = '../data/genres_original/blues/blues.00000.wav'

# Extrair features com nossa função
extracted = extract_features(audio_path)

print(f"Features extraídas: {len(extracted)}")
print(f"\nPrimeiros 5 valores extraídos:")
print(extracted[:5])

print(f"\nPrimeiros 5 valores do CSV:")
print(csv_features[:5])

print(f"\nDiferença absoluta (primeiros 5):")
print(np.abs(extracted[:5] - csv_features[:5]))
```

```python
# Comparação completa das 57 features
feature_names = get_feature_names()
diff = np.abs(extracted - csv_features)

print("Validação completa:")
print(f"{'Feature':<30} {'Extraído':>15} {'CSV':>15} {'Diferença':>15}")
print("-" * 78)
for name, ext, csv, d in zip(feature_names, extracted, csv_features, diff):
    status = "✓" if d < 1.0 else "⚠"
    print(f"{status} {name:<28} {ext:>15.6f} {float(csv):>15.6f} {d:>15.6f}")
```
