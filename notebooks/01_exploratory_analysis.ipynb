```python
# Importações necessárias
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Configurações de visualização
plt.style.use('ggplot')
sns.set_palette('husl')

print("Bibliotecas importadas com sucesso!")
```

```python
# Carregando o dataset de features
df = pd.read_csv('../data/features_30_sec.csv')

# Primeiras informações
print(f"Shape: {df.shape}")
print(f"\nColunas: {df.columns.tolist()}")
print(f"\nPrimeiras linhas:")
df.head()
```

```python
# Distribuição dos gêneros
print("Distribuição dos gêneros:")
print(df['label'].value_counts())

# Dados faltando
print(f"\nDados faltando: {df.isnull().sum().sum()}")
```

```python
# Visualizar distribuição
plt.figure(figsize=(10, 5))
df['label'].value_counts().plot(kind='bar', color=sns.color_palette('husl', 10))
plt.title('Distribuição de músicas por gênero')
plt.xlabel('Gênero')
plt.ylabel('Quantidade')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Separar features e label
X = df.drop(['filename', 'length', 'label'], axis=1)
y = df['label']

print(f"Features: {X.shape[1]}")
print(f"Exemplos: {X.shape[0]}")
print(f"\nGêneros únicos: {y.unique()}")
```

```python
# Correlação das features com o label (usando codificação numérica)
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
y_encoded = le.fit_transform(y)

# Top 10 features mais correlacionadas com o gênero
correlacoes = pd.DataFrame(X)
correlacoes['label_encoded'] = y_encoded

corr_com_label = correlacoes.corr()['label_encoded'].drop('label_encoded').abs().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
corr_com_label.head(15).plot(kind='bar', color=sns.color_palette('husl', 15))
plt.title('Top 15 features mais correlacionadas com gênero musical')
plt.xlabel('Feature')
plt.ylabel('Correlação absoluta')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

print("\nTop 10 features:")
print(corr_com_label.head(10))
```

```python
# Distribuição do spectral_bandwidth_mean por gênero
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='label', y='spectral_bandwidth_mean',
            hue='label', palette='husl', legend=False)
plt.title('Spectral Bandwidth por Gênero Musical')
plt.xlabel('Gênero')
plt.ylabel('Spectral Bandwidth Mean')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
